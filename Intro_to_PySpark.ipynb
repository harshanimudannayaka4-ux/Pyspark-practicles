```python
from pyspark.sql import SparkSession
spark = (SparkSession.builder
         .appName("Economic Analysis")
         .master("local[*]")
         .getOrCreate()
)
print("Pyspark started successfully")
```

```python
Econimc_data= [
    ("Sri Lanka",2021,7.0,5.1,3.5),
    ("Sri Lanka",2022,46.4,4.7,-7.3),
    ("Sri Lanka",2023,17.4,4.7,-2.3),
    ("India",2021,5.1,6.2,9.7),
    ("India",2022,6.7,7.3,7.0),
    ("India",2023,5.7,4.8,8.2),
    ("Bangladesh",2021,5.5,5.2,6.9),
    ("Bangladesh",2022,7.7,4.7,7.1),
    ("Bangladesh",2023,9.9,4.5,5.8)
]
columns = [
    "country",
    "year",
    "inflation_rate",
    "unemployment_rate",
    "gdp_growth"
]
df=spark.createDataFrame(Econimc_data,columns)
df.show()
```

```python
df.show(5)
```

```python
print(df.columns)
```

```python
print(df.count())
```

```python
df.printSchema()
```

```python
df.select(
    "country",
    "year",
    "inflation_rate"
).show()
```

```python
from pyspark.sql.functions import col
df.select(
    col("country"),
    col("inflation_rate")
).show()
```

The select() operation creates a new DataFrame containing the selected columns

```python
high_inflation = df.filter(
    col("inflation_rate")>10
)
high_inflation.show()
```

```python
df.filter(
    col("year")==2023
).show()
```

```python
df.filter(
    col("country")=="Sri Lanka"
).show()
```

**Multiple** **Conditions**

```python
df.filter(
    (col("inflation_rate")>10) &
    (col("gdp_growth")<10)
).show()
```

```python
df.orderBy(
    col("inflation_rate").asc()
).show()
```

```python
df.orderBy(
    col("inflation_rate").desc()
).show()
```
